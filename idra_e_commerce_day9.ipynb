```python
import pandas as pd

'''
1.Create a Processed E-commerce Dataset using the three provided CSV files: Orders, Customers, and Products in the "Study Material" tab in the LMS.
2.Load the datasets using Pandas and combine the related information using merge(), while using concat() to demonstrate combining DataFrames.
3.Use apply() to create or transform useful columns, and use DateTime operations to convert the order date and extract information such as the month, day, or day of the week.
4.Organize the resulting information into a clean, meaningful DataFrame and export the final processed dataset as a CSV file.
5.Upload the notebook, the input datasets and the final processed dataset on a GitHub repository before submitting your work.
'''
# Data Concatenation

customers = pd.read_csv('Day9_Customers.csv')
products = pd.read_csv('Day9_Products.csv')
orders = pd.read_csv('Day9_Orders.csv')
```

```python
#Simple concatenation (combining tables horizontally)
combined = pd.concat([orders, customers, products], ignore_index = True)
combined
```

```python
#Left merge
left = pd.merge(orders, customers, on = 'Customer_ID', how = 'left')
left
```

```python
#right merge
right = pd.merge(orders, products, on = 'Product_ID', how = 'left')
right
```

```python
# Inner merge and display based on common customer_ID and Product_ID

# merge orders and customers on 'Customer_ID'
merged1 = pd.merge(orders, customers, on = 'Customer_ID')

# merge merged1(orders,customers) and products on 'Product_ID'
merged2 = pd.merge(merged1, products, on = 'Product_ID')

# we get records where we have both Product_ID and Customer_ID
merged2
```

```python
# .apply() to calculate Total_Sales
def calculate_total_sales(row):
    return row['Unit_Price'] * row['Quantity']

merged2['Total_Sales'] = merged2.apply(calculate_total_sales, axis=1)
```

```python
# Convert 'Order_Date' to datetime objects
merged2['Order_Date'] = pd.to_datetime(merged2['Order_Date'])
```

```python
# Extract month, day, and day of the week
merged2['Order_Month'] = merged2['Order_Date'].dt.month
merged2['Order_Day'] = merged2['Order_Date'].dt.day
merged2['Order_DayOfWeek'] = merged2['Order_Date'].dt.dayofweek
```

```python
# Display the resulting DataFrame
display(merged2.head(20))
```

```python
# Export the final processed dataset as a CSV file
merged2.to_csv('processed_ecommerce_dataset.csv', index=False)
print("Processed dataset exported to 'processed_ecommerce_dataset.csv'")
```
